This notebook demonstrates Python programming and basic statistical analysis.

## 1. Dataset
A small list of student scores is used for demonstration.

```python
scores = [45, 72, 81, 39, 66, 90, 72]
scores
```

## 2. Basic Statistics
Calculate mean, median, mode, standard deviation, minimum, maximum, and number of passing scores.

```python
import statistics

mean_score = statistics.mean(scores)
median_score = statistics.median(scores)
mode_score = statistics.mode(scores)
std_dev = statistics.stdev(scores)

print("Mean:", mean_score)
print("Median:", median_score)
print("Mode:", mode_score)
print("Standard deviation:", round(std_dev, 2))
print("Minimum:", min(scores))
print("Maximum:", max(scores))
print("Number passed:", sum(s >= 50 for s in scores))
```

## 3. Python Data Structures
The example below uses a dictionary to store summary information.

```python
summary = {
    "count": len(scores),
    "mean": mean_score,
    "median": median_score,
    "highest": max(scores),
    "lowest": min(scores),
    "passed": sum(s >= 50 for s in scores)
}
summary
```

## 4. Interpretation
- The scores are numerical data.
- Mean and median describe the center of the distribution.
- Standard deviation describes the spread of the scores.
- A simple condition is used to count passing scores.
- The analysis should not be generalized beyond the dataset without additional evidence.
